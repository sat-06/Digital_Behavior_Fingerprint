```python
# 16. Identify ID columns
id_columns = ["subject", "sessionIndex", "rep"]
print("ID Columns:", id_columns)

# 17. Identify behavioural feature columns
behaviour_features = [col for col in df.columns if col not in id_columns]
print("\nBehavioural Features:")
print(behaviour_features)
print("Number of behavioural features:", len(behaviour_features))

# 18. Distribution of H.period
plt.figure(figsize=(8, 5))
sns.histplot(df["H.period"], kde=True)
plt.title("Distribution of H.period")
plt.xlabel("H.period")
plt.ylabel("Frequency")
plt.show()

# 19. Distribution of a few timing features
features = ["H.period", "DD.period.t", "UD.period.t"]

for feature in features:
    plt.figure(figsize=(8, 5))
    sns.histplot(df[feature], kde=True)
    plt.title(f"Distribution of {feature}")
    plt.xlabel(feature)
    plt.ylabel("Frequency")
    plt.show()

# 20. Compare H.period across different subjects
subjects = df["subject"].unique()[:5]

plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df[df["subject"].isin(subjects)],
    x="subject",
    y="H.period"
)
plt.title("H.period Across Different Subjects")
plt.xlabel("Subject")
plt.ylabel("H.period")
plt.show()

# 21. Correlation matrix
correlation = df[behaviour_features].corr()

# 22. Correlation heatmap
plt.figure(figsize=(16, 12))
sns.heatmap(correlation, cmap="coolwarm")
plt.title("Correlation Between Behavioural Features")
plt.show()
```
